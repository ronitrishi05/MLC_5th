```python
res=df.corr(method='pearson')
print("Corelation matrix:\n",res)
```

```python
x = df.drop("Final_Score", axis=1)
y = df["Final_Score"] 
correlations = x.corrwith(y)
print(correlations)
```

```python
tar_corr=correlations.abs().sort_values(ascending=False)
print("Target variable correlation:\n",tar_corr)
print("The most correlated feature with the target variable is:",tar_corr.index[0])
```
