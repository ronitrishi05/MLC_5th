```python
x = np.linspace(-2,2,15)
y=2*x+3
z=x*x+1
plt.subplot(1,4,1)
plt.scatter(x,y,color='r')
plt.subplot(1,4,2)
plt.scatter(x,z,color='r')

plt.subplot(1,4,3)
plt.plot(x,y,marker='.')
plt.subplot(1,4,4)
plt.plot(x,z,marker='.')
```

```python
x = np.linspace(-2,2,15)
y=2*x+3
z=x*x+1
plt.plot(x,y,label="linear functioj")
plt.plot(x,z,label="Non linear")
plt.legend(loc='lower left')

plt.xlim(-1,1)
plt.ylim(0,4)
```
