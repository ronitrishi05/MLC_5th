```python
import numpy as np
import pandas as pd
x=np.random.randint(0,18,20)
y=np.where(x>3,1,0)
df=pd.DataFrame({'x':x,'y':y})
x=df[['x']]
y=df['y']

from sklearn import linear_model
lr=linear_model.LinearRegression()
lr.fit(x,y)
ypr=lr.predict(x)
print(ypr)
cp1=np.where(ypr>0.5,1,0)

from sklearn.metrics import accuracy_score
print("Accuracy:",accuracy_score(y,cp1))

```

```python
from sklearn.linear_model import LogisticRegression
lr1=LogisticRegression()
lr1.fit(x,y)
ypr1=lr1.predict(x)
print("Accuracy:",accuracy_score(y,ypr1))
print("Classificaion prob. :",lr1.predict_proba(x))
```
