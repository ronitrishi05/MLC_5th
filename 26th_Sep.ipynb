```python
import numpy as np
import matplotlib.pyplot as plt
import random
import pandas as pd
from scipy import stats 

from sklearn import linear_model
```

```python
#define th dataset
x=[[1],[2],[3],[4]]
y=[3,5,7,9]
x2=[[1],[2],[3],[4]]
```

```python
#perform linear regression 
lr=linear_model.LinearRegression()
lr.fit(x,y)#entire data for training
#perfrom prediction
y_pred=lr.predict(x2)
print(y_pred)
```

```python
#findout coeeficient and intercept
cf=lr.coef_
intc=lr.intercept_
print("Coefficient :",cf)
print("Intercept:",intc)
```

```python
#evaluate RSS
from sklearn import metrics
print("RSS:",(np.sum(y-y_pred)**2).round(2))
print("MAE",metrics.mean_absolute_error(y,y_pred).round(2));
print("MSE",metrics.mean_squared_error(y,y_pred).round(2));
print("RMSE",np.sqrt(metrics.mean_squared_error(y,y_pred).round(2)));
```

```python
plt.scatter(x,y,color="red",label="Actual")
plt.scatter(x,y_pred,color="green",label="Predicted")
plt.plot(x,y,label="Linear regression line")

plt.legend()
equ=f"y={lr.coef_[0].round(2)}*x+{lr.intercept_.round(2)}"
plt.text(3,4,equ,c="r")
plt.show()
```

```python
xo=np.random.randint(0,10,4)
x3=xo.reshape(4,1)
y_3=lr.predict(x3)
print(y_3)
```

```python
xi=[1,2,3,4,5]
xj=[10,20,30,40,50]
y4=[13,25,37,49,51]
df=pd.DataFrame({"x1":xi,"x2":xj,"y":y4})
x_input=df[['x1','x2']]
y=df["y"]
```

```python
#model training and testing
from sklearn import linear_model
lr=linear_model.LinearRegression()
lr.fit(x_input,y)
ypred=lr.predict(x_input)
print(ypred)
```

```python
#coef and intercept
print("Coefficient :",lr.coef_.round(2))
print("Intercept:",lr.intercept_.round(2))
```

```python
#find the eroor
print("RSS:",(np.sum((y-ypred)**2)).round(2))
print("MAE",metrics.mean_absolute_error(y,ypred).round(2));
print("MAE",round(np.mean(np.abs(y-ypred)),2))
print("MSE",metrics.mean_squared_error(y,ypred).round(2))
print("MSE",np.mean((y-ypred)**2))
print("RMSE",np.sqrt(metrics.mean_squared_error(y,ypred)))
print("RMSE",np.sqrt(np.mean(np.abs(y-ypred)**2)))
```
