```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import sklearn 
import scipy
print("Success")
```

```python
df=pd.read_csv("Student_Performance_Prediction.csv")
df
```

```python
from sklearn import linear_model
lr=linear_model.LinearRegression()
```

```python
x = df.drop("Final_Score", axis=1)
y = df["Final_Score"] 
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
lr.fit(xtrain,ytrain)
```

```python
ypred=lr.predict(xtest)
```

```python
from sklearn import metrics
RSS=(np.sum(ytest-ypred)**2)
print("RSS:",RSS)
print("MAE",metrics.mean_absolute_error(ytest,ypred));
print("MSE",metrics.mean_squared_error(ytest,ypred));
print("RMSE",np.sqrt(metrics.mean_squared_error(ytest,ypred)));
print("R2 value:",metrics.r2_score(ytest,ypred))
```

```python
from scipy import stats
res=df.corr(method='pearson')
print("Corelation matrix:\n",res)
```

```python
correlations = x.corrwith(y)
print(correlations)
```

```python
strongest_features = correlations.abs().sort_values(ascending=False)
print("Features ranked by relationship strength (absolute value):")
print(strongest_features)
```
