```python
import numpy as np
import matplotlib.pyplot as plt
import random
import pandas as pd
from scipy import stats 
```

```python
ns = 100  # number of samples
sd = np.random.randint(1, 101, ns)
ts = 4 * sd + np.random.normal(0, 20, ns)
df = pd.DataFrame({
    "StudyHour": sd,
    "TestScore": ts
})
print("Complete Dataset:")
print(df)
```

```python
print("\nNumber of samples:", df.shape[0])
print("Number of features:", 1)
print("Number of target variables:", 1)
print("Feature name:", "StudyHour")
print("Target name:", "TestScore")
print("Shape of dataset:", df.shape)
print("Data type of feature:", df["StudyHour"].dtype)
```

```python
print("\nStatistics:")
print(df.describe())
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
plt.scatter(df["StudyHour"], df["TestScore"])

plt.title("Study Hours vs Test Score")
plt.xlabel("Study Hours")
plt.ylabel("Test Score")
```
