```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
A = np.array([ [2, -4], [-6, 8]])
B = np.array([ [1, -3], [-5, -7]])

print("A+B:")
print(A + B)
print("A-B:")
print(A - B)
print("A.B:")
print(A @ B)
print("Transpose of A:")
print(A.T)
print("Determinant of A:")
print(np.linalg.det(A))
print("Inverse of A:")
print(np.linalg.inv(A))
```
