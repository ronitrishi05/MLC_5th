```python
MARKS = np.array([ [80, 75, 90], [65, 70, 72], [90,88, 95], [55, 60, 58]])
print("Average of each student marks: ", np.mean(MARKS, axis=1))
print("Highest student Average marks: ", np.max(np.mean(MARKS, axis=1)))
print("Average of each subject marks: ", np.mean(MARKS, axis=0))
print("Highest subject Average marks: ", np.max(np.mean(MARKS, axis=0)))
print("student averages MORE THAN 75:",np.where(np.mean(MARKS, axis=1) > 75)[0])
```
