```python
import numpy as np
import matplotlib.pyplot as plt
import random
import pandas as pd
from scipy import stats 
```

```python
data={
    'product':['laptop','Smartphone','Monitor','Headphone','Mouse'],
    'Price':[200000,180000,15000,32000,5000],
    'Sales':[150000,120000,10000,12000,1000],
    'profit':[50000,20000,5000,5000,2000]
}
df=pd.DataFrame(data)
plt.scatter(df['Price'],df['Sales'],c=df['profit'],s=df['profit'] / 100, cmap='viridis')
for i in range(len(df)):
    plt.annotate(df['product'][i], (df['Price'][i], df['Sales'][i]), xytext=(8,8), textcoords='offset points')
```
