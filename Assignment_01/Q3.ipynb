```python
import pandas as pd
data={
    'product':['laptop','Smartphone','Monitor','Headphone','Mouse'],
    'Price':[200000,180000,15000,32000,5000],
    'Sales':[150000,120000,10000,12000,1000],
    'profit':[50000,20000,5000,5000,2000]
}
df=pd.DataFrame(data)
print("Dataframe:\n",df)
print("\nColumn Name:",df.columns)
print("\nNumber Of Rows & columns :",df.shape)
print("\nFirst 3 Rows:\n",df.head(3))
print("\nOnly Sales And Profit Column:\n",df[['Sales','profit']])
print("\nAverage Sales:", df['Sales'].mean())
print("Maximum Sales:", df['Sales'].max())
print("Minimum Sales:", df['Sales'].min())
print("\nProducts with Profit > 4000:\n",df[df['profit'] > 4000]['product'])
df.sort_values('profit',ascending=False)
print("Dataframe sorted by Profit (Descending):\n",df)
```
