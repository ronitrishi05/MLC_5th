```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Products': ['Laptop', 'Mobile', 'Tablet', 'Headphones', 'Smartwatch'],
    'Price': [60000, 25000, 30000, 5000, 8000],
    'Sales': [100, 250, 150, 500, 300],
    'Profit': [8000, 5000, 7000, 2000, 4500]
}

df = pd.DataFrame(data)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['Price'], df['Sales'], marker='o', label='Products')
High_ind = df['Sales'].idxmax()
plt.scatter(
    df.loc[High_ind,'Price'],
    df.loc[High_ind,'Sales'],
    marker='o',
    color='red',
    s=150,
    label='Highest Sales'
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df['Products'],
    df['Sales'],
    marker='o',
    linestyle='-',
    label='Sales'
)
plt.plot(
    df['Products'],
    df['Profit'],
    marker='s',
    linestyle='--',
    label='Profit'
)
```

```python
plt.figure(figsize=(8, 5))

bars = plt.bar(df['Products'], df['Profit'])

plt.bar_label(bars, padding=1)

plt.xlabel('Products')
plt.ylabel('Profit')
plt.title('Profit by Product')

plt.show()
```

```python

profit_margin = df['Profit'] / df['Sales'] * 100
plt.figure(figsize=(8, 5))
bars = plt.bar(df['Products'], profit_margin)
plt.bar_label(bars, padding=1)
```

```python
plt.figure(figsize=(7, 7))

plt.pie(
    df['Sales'],
    labels=df['Products'],
    autopct='%1.1f%%'
)
plt.title('Percentage Contribution of Each Product to Total Sales')
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Sales'], bins=5, edgecolor='black')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.title('Distribution of Sales')
```
