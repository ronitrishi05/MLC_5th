```python
#Q1- suppose cpu utiliz of a cloud server is ranges from 10% to 35% in 50 diffeent time stamp 
# prepare a dtaframe to represent the dat in termw of time stramp and cpu utiliz. 
import numpy as np
import matplotlib.pyplot as plt
import random
import pandas as pd
data =np.random.randint(10,30,50)
print(data)
df=pd.DataFrame({'cpu':data})
df['time']=df.index
df
```

```python
#q2 disopplay the min and max value of cpu utiliz present in the dataframe . 
print("min",df.min())
print("Max",df.max())
```

```python
#q3- display the valu pf low cpu utilz when the value is below 15%
print(df[df['cpu']<15])
```

```python
#q4- plot a bar graph proper labe; and title and sisplay cpu utilz over it 
plt.bar(df.time,df.cpu)
plt.show() 
```
