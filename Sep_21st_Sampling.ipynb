```python
import numpy as np
import matplotlib.pyplot as plt
import random
import pandas as pd
from scipy import stats 
```

```python

a=np.random.randint(20,46,30)
df=pd.DataFrame({'temp':a})
df
```

```python
df['class']=np.where(df.temp>=40 ,'H','L')
print("Number of H :\n",df['class'].value_counts())
```

```python
rs=df.sample(frac=0.1)
rs2=df.groupby('class').sample(frac=0.1)
print(rs2['class'].value_counts())
```

```python
x=np.random.randint(-10,10,100)
y1=2*x+3;
y2=np.random.uniform(-3,3,100)
plt.subplot(1,2,1)
plt.scatter(x,y1)
plt.subplot(1,2,2)
plt.scatter(x,y2)
plt.show()
```

```python
(corr,p1)=stats.pearsonr(x,y1)
(corr2,p2)=stats.pearsonr(x,y2)
print("correlation:",corr,"\nP-values:",p1)
print("correlation 2:",corr2,"\nP-values 2:",p2)
```
